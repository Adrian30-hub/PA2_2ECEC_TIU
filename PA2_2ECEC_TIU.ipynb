#### A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
x = np.mean(X)
x
```

```python
a = np.std(X)
a
```

```python
X_normalized = (X-x)/a
X_normalized
```

```python
X
```

```python
X_normalized
```

```python
print("Mean:", round(X_normalized.mean()))
```

```python
print("Standard Deviation:", round(X_normalized.std()))
```

```python
np.save("X_normalized", X_normalized)
```

#### B. CUBES DIVISIBLE BY 4 PROBLEM

```python
C = (np.linspace(1,100,100) ** 3).reshape(10,10)
C
```

```python
C.shape
```

```python
div_by_4 = C[C % 4 == 0]
div_by_4
```

```python
div_by_4.shape
```

```python
np.save("div_by_4", div_by_4)
```

#### C. ABOVE-MEAN SQUARES PROBLEM

```python
S = (np.linspace(1,36,36) ** 2).reshape(6,6)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S > S_mean]
above_mean
```

```python
above_mean.shape
```

```python
np.save("above_mean", above_mean)
```
